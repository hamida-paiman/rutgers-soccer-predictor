```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import pandas as pd

# --- 1. Split by season (time-based) ---
train = data_clean[data_clean['Season'].isin([2023, 2024])].copy()
test = data_clean[data_clean['Season'] == 2025].copy()

# --- 2. Encode the target (Result: W/L/T -> numbers) ---
outcome_map = {'L': 0, 'T': 1, 'W': 2}
train['outcome_encoded'] = train['Result'].map(outcome_map)
test['outcome_encoded'] = test['Result'].map(outcome_map)

# --- 3. Build features (X) ---
# One-hot encode categorical columns. pd.get_dummies on train and test 
# SEPARATELY can create mismatched columns, so combine first, encode, then split back.
combined = pd.concat([train, test])
combined_encoded = pd.get_dummies(combined, columns=['At', 'Opponent'])

train_encoded = combined_encoded[combined_encoded['Season'].isin([2023, 2024])]
test_encoded = combined_encoded[combined_encoded['Season'] == 2025]

feature_cols = [col for col in combined_encoded.columns if col.startswith('At_') or col.startswith('Opponent_')]

X_train = train_encoded[feature_cols]
y_train = train_encoded['outcome_encoded']
X_test = test_encoded[feature_cols]
y_test = test_encoded['outcome_encoded']

# --- 4. Fit Logistic Regression ---
logreg = LogisticRegression(max_iter=1000)
logreg.fit(X_train, y_train)
logreg_preds = logreg.predict(X_test)

print("Logistic Regression Accuracy:", accuracy_score(y_test, logreg_preds))
print(confusion_matrix(y_test, logreg_preds))
print(classification_report(y_test, logreg_preds, target_names=['L', 'T', 'W']))

# --- 5. Fit Decision Tree ---
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X_train, y_train)
tree_preds = tree.predict(X_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, tree_preds))
print(confusion_matrix(y_test, tree_preds))
print(classification_report(y_test, tree_preds, target_names=['L', 'T', 'W']))
```

```python
r_2026 = rDataset[rDataset['season'] == 2026]
w_2026 = wDataset[wDataset['season'] == 2026]

rutgers_gpg = float(r_2026[r_2026['Stat'] == 'Goals Per Game']['Rutgers'].values[0])
rutgers_ga = float(r_2026[r_2026['Stat'] == 'Goals Per Game']['Opponents'].values[0])

washington_gpg = float(w_2026[w_2026['Stat'] == 'Goals Per Game']['Washington'].values[0])
washington_ga = float(w_2026[w_2026['Stat'] == 'Goals Per Game']['Opponents'].values[0])

print(rutgers_gpg, rutgers_ga, washington_gpg, washington_ga)
```

```python
home_boost = 1.10
away_penalty = 0.90

rutgers_expected_goals = ((rutgers_gpg * away_penalty) + washington_ga) / 2
washington_expected_goals = ((washington_gpg * home_boost) + rutgers_ga) / 2

print(f"Rutgers expected goals: {rutgers_expected_goals:.2f}")
print(f"Washington expected goals: {washington_expected_goals:.2f}")
```

```python
from scipy.stats import poisson

max_goals = 6
rutgers_dist = [poisson.pmf(i, rutgers_expected_goals) for i in range(max_goals + 1)]
washington_dist = [poisson.pmf(i, washington_expected_goals) for i in range(max_goals + 1)]

win_prob = draw_prob = loss_prob = 0
for r in range(max_goals + 1):
    for w in range(max_goals + 1):
        p = rutgers_dist[r] * washington_dist[w]
        if r > w:
            win_prob += p
        elif r == w:
            draw_prob += p
        else:
            loss_prob += p

print(f"Rutgers Win: {win_prob:.1%}")
print(f"Draw: {draw_prob:.1%}")
print(f"Rutgers Loss: {loss_prob:.1%}")
```
